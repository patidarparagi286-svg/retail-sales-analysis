**Business Analysis**

```python

print("Total Sales:", df["Sales"].sum())
print("Total Profit:", df["Profit"].sum())
print("Total Quantity Sold:", df["Quantity"].sum())
print("Total Orders:", df["Order ID"].nunique())
```

TOPIC : Category-wise Sales Analysis

```python
category_sales = df.groupby("Category")["Sales"].sum()
print(category_sales)
```

```python
import matplotlib.pyplot as plt
category_sales.plot(kind="bar")
plt.title("Sales by Product Category")
plt.xlabel("Category")
plt.ylabel("Total Sales")
plt.xticks(rotation=0)
plt.show()
```

TOPIC : Category-wise Profit Analysis

```python
category_profit = df.groupby("Category")["Profit"].sum()

print(category_profit)
```

```python
category_profit.plot(kind="bar")
plt.title("Profit by Product Category")
plt.xlabel("Category")
plt.ylabel("Total Profit")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

TOPIC : Region-wise Sales Analysis

```python
region_sales = df.groupby("Region")["Sales"].sum()

print(region_sales.sort_values(ascending=False))
```

```python
region_sales.sort_values(ascending=False).plot(kind="bar")

plt.title("Sales by Region")
plt.xlabel("Region")
plt.ylabel("Total Sales")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

TOPIC : Sub-Category-wise Profit Analysis

```python
subcategory_profit = df.groupby("Sub-Category")["Profit"].sum()

print(subcategory_profit.sort_values(ascending=False))
```

```python
top_10_profit = subcategory_profit.sort_values(ascending=False).head(10)

top_10_profit.plot(kind="bar", figsize=(10, 5))

plt.title("Top 10 Sub-Categories by Profit")
plt.xlabel("Sub-Category")
plt.ylabel("Total Profit")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

TOPIC : Monthly Sales Trend Analysis

```python
#coverting date column into datetime
df["Order Date"] = pd.to_datetime(df["Order Date"])
```

```python
df["Month"] = df["Order Date"].dt.month

monthly_sales = df.groupby("Month")["Sales"].sum()

print(monthly_sales)
```

```python
monthly_sales.plot(kind="line", marker="o", figsize=(10, 5))

plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Total Sales")
plt.xticks(range(1, 13))
plt.grid(True)
plt.tight_layout()
plt.show()
```

TOPIC : Monthly Profit Analysis

```python
monthly_profit = df.groupby("Month")["Profit"].sum()

print(monthly_profit)
```

```python
monthly_profit.plot(kind="line", marker="o", figsize=(10, 5))

plt.title("Monthly Profit Trend")
plt.xlabel("Month")
plt.ylabel("Total Profit")
plt.xticks(range(1, 13))
plt.grid(True)
plt.tight_layout()
plt.show()
```

TOPIC : Sales vs. Profit Analysis

```python
plt.figure(figsize=(8, 5))

plt.scatter(df["Sales"], df["Profit"], alpha=0.5)

plt.title("Sales vs Profit")
plt.xlabel("Sales")
plt.ylabel("Profit")
plt.axhline(y=0, linestyle="--")
plt.tight_layout()
plt.show()
```

TOPIC : Correlation between Sales and Profit


```python
correlation = df["Sales"].corr(df["Profit"])

print("Sales-Profit Correlation:", correlation)
```

TOPIC : Profit mergin

```python
df["Profit_Margin"] = (df["Profit"] / df["Sales"]) * 100

print(df[["Category", "Profit_Margin"]].head())
```

TOPIC : Discount vs Profit Analysis

```python
import matplotlib.pyplot as plt

plt.scatter(df["Discount"], df["Profit"])
plt.xlabel("Discount")
plt.ylabel("Profit")
plt.title("Discount vs Profit")
plt.show()
```
